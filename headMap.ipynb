```python
data = sns.load_dataset("tips")
data
```

```python
plt.figure(figsize=(5,5))
sns.barplot(x="sex", y="size", data=data)
```

```python
sns.countplot(x="sex",data=data,hue="smoker",palette="bwr",saturation=.1)
```

```python
sns.violinplot(x="day",y="total_bill", data=data, hue="time",linewidth=2,palette="Oranges_r")
```

```python
sns.violinplot(x="time",y="total_bill", data=data, order=["Dinner","Lunch"],split=True)
```

```python
sns.violinplot(x="time",y="total_bill",data=data,inner="quartile")
```

```python
sns.violinplot(x=data["total_bill"])
```

```python
xy=sns.load_dataset("tips")
sns.pairplot(xy)
```

```python
sns.stripplot(x="day", y="total_bill", data=data,hue="sex",palette="Oranges_r",jitter=1,edgecolor="r",size=1)
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="total_bill",y="day",data=data,hue="sex",color="g",palette="plasma")
```

```python
sns.set(style="whitegrid")

sns.boxplot(
    data=data,
    x="total_bill",
    hue="sex",
    palette="plasma",
    linewidth=1,
    orient="h",
    showmeans=True
)

plt.show()
```

```python
sns.set_style("darkgrid")
sns.set_context("poster", font_scale=0.5)
sns.catplot(
    x="size",
    # y="tip",
    data=data,
    # kind="strip"
    # orient="h"
)
# plt.grid()
```

```python
fg =sns.FacetGrid(data,col="day",hue="sex")
fg.map(plt.scatter,"tip","total_bill" ).add_legend()
fg
```
