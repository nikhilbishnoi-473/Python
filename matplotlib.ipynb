```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 33, 45, 50]

plt.plot(x, y)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.title("total number")
plt.xlabel("y axis number")
plt.ylabel("x axis label")
plt.plot(x, y, color='green', marker='o', linestyle='dashed',
     linewidth=2, markersize=12)
plt.show()
```

```python
 
x =[1,2,3,4,5]
y1 =[10,23,34,43, 50]
y2 =[20,24,35,44,56]
plt.plot(x,y1,label='Sales 2024', color='y',marker='.', linestyle='dashed',linewidth=2,markersize=12)
plt.plot(x,y2,label='Sales 2025')
plt.title("total number ")
plt.xlabel("x-axis label")
plt.ylabel("y-axis number")
plt.legend()
plt.show()


```

```python
# bar chat
x = ["a","b","c","d","e"]
y = [20,30,35,50,45]
plt.bar(x,y,color='r'  )
plt.show()
```

```python
import random
plt.Figure(figsize=(4,4))
data = [random.randint(1,50) for _ in range(500)]
plt.hist(data, bins=5)
plt.show()
```

```python
x = ["A","B","C","D","E"]
y = [10,40,50,20,34]
plt.pie(y, labels=x,autopct='%1.1f%%',startangle=67)
plt.title("pie chart")
plt.show()
```

```python
import matplotlib.pyplot as plt
x1 = [10,20,30,40,50]
x2 = [10,20,30,40,50]
y1 = [10, 20, 30, 40, 50]
y2 = [10, 20, 30, 40, 50]
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.bar(x1,x2)
plt.title("one bar chart")
plt.xlabel("x-axis bar")
plt.ylabel("x-axis bar")
plt.subplot(1,2,2)
plt.scatter(y1,y2)
plt.title("Scatter chart")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()


```

```python
import pandas as pd
data = {
    "Month":["jan","feb","mar", "apr"],
    "Salary":[120000,1400000,2000000,400000]
}
df = pd.DataFrame(data)
plt.bar(df["Month"],df["Salary"])
plt.savefig("python_graph.png")
plt.show()
```
