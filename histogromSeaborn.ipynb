```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
var = sns.load_dataset("penguins").head(200)
var
```

```python
sns.histplot(
    var["flipper_length_mm"],
    kde=True,
    color="g",
    log_scale=True,
    # multiple='layer'
)
sns.rugplot(x=var["flipper_length_mm"])

plt.show()
```

```python
m={"Male":"o","Female": "*"}
sns.scatterplot(x="bill_length_mm", y="bill_depth_mm", data=var, hue="sex", style="sex", size="sex",sizes=[40,90], palette="PuBu",alpha=1, markers=m)
plt.show()
```
